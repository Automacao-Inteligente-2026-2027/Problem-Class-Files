#### 1.2. Plot a correlation matrix of the columns (with a statistical significance of 0.1%) and a scatter plot of medv(lstat)

```python
## Import pearsonr function from scipy -- calculate the correlation and p-value between two columns

## Calculate the correlations between the columns
df_corrs = df.corr(method='pearson')

## Calculate the p-value, as the second element of the response from the pearsonr function.
pval = df.corr(method=lambda x, y: scipy.stats.pearsonr(x, y)[1])

## Establish the mask, to hide values without a given statistical significance
ptg_stat_sig = 0.1 / 100
mask = pval > ptg_stat_sig

## Plot the correlation matrix using seaborn's heatmap function
plt.subplots(figsize=(15, 15))
heatmap = sns.heatmap(df_corrs, mask=mask, square=True, cmap='coolwarm', annot=True)
```

```python
sns.scatterplot(data=df, x='lstat', y='medv')
```

#### 1.3. Fit a linear model considering *medv* and *lstat*, with *medv* being the response and *lstat* the predictor. Use the whole dataset -- do not divide in train and test datasets. Calculate the coefficients and the $R^2$

(Suggestion: use `sklearn.linear_model.LinearRegression` and `sklearn.metrics.r2_score`)

```python
## Assign the predictor and response to variables. To fit the data, the pandas dataframe must be converted into a numpy
## array in a matricial shape, even if with a single column
X = df['lstat'].values.reshape(-1, 1)
y = df['medv'].values.reshape(-1, 1)

## Initialize the model as a LinearRegression() function, and fit the data to it
regr = sklearn.linear_model.LinearRegression()
regr.fit(X, y)

## Print the interception and coefficient
print('medv = {:.4f} + {:.4f} * lstat'.format(regr.intercept_[0], regr.coef_[0][0]))

## To calculate the R^2 score, the prediction of medv is calculate using the predict method.
medv_pred = regr.predict(X)
r2_value = sklearn.metrics.r2_score(y, medv_pred)
print('R^2: {:.4f}'.format(r2_value))
```

#### 1.4. Plot the fitted linear regression in the scatter plot from exercise 3.

```python
## Determine limits (min/max) of lstat and calculate the corresponding value of medv. Important: this can be done between
## only two points because this is a linear regression.
lstat_vals = np.array([min(X), max(X)])
medv_vals = regr.intercept_[0] + regr.coef_[0][0] * lstat_vals

## Plot the scatterplot and the regression line
fig, ax = plt.subplots(figsize=(12, 6))
sns.scatterplot(data=df, x='lstat', y='medv', color='red')
ax.plot(lstat_vals, medv_vals, 'b-')
```

#### 1.5. Plot the residuals against the fitted values

```python
## Calculate the residuals as the difference between the actual response and the predicted response
residuals_y = y - medv_pred
residuals_x = medv_pred

fig, ax = plt.subplots(figsize=(12, 6))
ax.plot(residuals_x, residuals_y, '.b')
ax.grid(visible=True)
ax.set_xlabel('medv predicted')
ax.set_ylabel('medv residuals')
```

#### 1.6. Fit a linear model with *medv* as the response and *lstat* and *age* as the predictors. Calculate the coefficients and the $R^2$

```python
## Assign the predictor and response to variables. To fit the data, the pandas dataframe must be converted into a numpy
## array in a matricial shape, even if with a single column
cols = ['lstat', 'rm']
X = df[cols].values
y = df['medv'].values.reshape(-1, 1)

## Initialize the model as a LinearRegression() function, and fit the data to it
regr = sklearn.linear_model.LinearRegression()
regr.fit(X, y)

## Print the interception and coefficient
print('medv = {:.4f} + {:.4f} * lstat + {:.4f} * age'.format(regr.intercept_[0], regr.coef_[0][0], regr.coef_[0][1]))

## To calculate the R^2 score, the prediction of medv is calculate using the predict method.
medv_pred = regr.predict(X)
r2_value = sklearn.metrics.r2_score(y, medv_pred)
print('R^2: {:.4f}'.format(r2_value))
```

#### 1.7. Fit a linear model with *medv* as the response and all the other columns as the predictors. Calculate the coefficients and the $R^2$

#### 1.8. Fit a linear model with *medv* as the response and *lstat* and *age* as the predictors. Consider interactions terms. Calculate the coefficients and the $R^2$

(Suggestion: use `sklearn.preprocessing.PolynomialFeatures` for the establishing interaction elements)

```python
# Prepare the inputs and outputs
cols = ['lstat', 'age']
X = df[cols].values
y = df['medv'].values.reshape(-1, 1)

# Use PolynomialFeatures with interaction_only to transform the input matrix
poly = sklearn.preprocessing.PolynomialFeatures(interaction_only=True, include_bias=False)
X = poly.fit_transform(X)

# Fit the data to the linear model
regr = sklearn.linear_model.LinearRegression()
regr.fit(X, y)

# Print the results
print('medv = {:.4f} + {:.4f} * lstat + {:.4f} * age + {:.4f} * lstat * age'.format(
    regr.intercept_[0], regr.coef_[0][0], regr.coef_[0][1], regr.coef_[0][2]
))

# Calculate the R^2 score between the actual output and the predicted one
medv_pred = regr.predict(X)
r2_value = sklearn.metrics.r2_score(y, medv_pred)
print('R^2: {:.4f}'.format(r2_value))
```

#### 1.10. Fit a linear model with *medv* as the response and *lstat* as the predictor. Consider terms of second degree. Calculate the coefficients and the $R^2$

(Note that selecting `interaction_only=True` in the PolynomialFeatures function does not allow quadratic terms; here those are required)

```python
# Prepare input and output datasets
X = df['lstat'].values.reshape(-1, 1)
y = df['medv'].values.reshape(-1, 1)

# Apply the polynomial transformation to the input dataset -- consider 2nd degree and without interaction
poly = sklearn.preprocessing.PolynomialFeatures(degree=2, interaction_only=False, include_bias=False)
X = poly.fit_transform(X)

# Fit the linear model
regr = sklearn.linear_model.LinearRegression()
regr.fit(X, y)

# Print the fitted model coefficients
print('medv = {:.4f} + {:.4f} * lstat + {:.4f} * lstat^2'.format(
    regr.intercept_[0], regr.coef_[0][0], regr.coef_[0][1]
))

# Calculate and print the R^2 score
medv_pred = regr.predict(X)
r2_value = sklearn.metrics.r2_score(y, medv_pred)
print('R^2: {:.4f}'.format(r2_value))
```

#### 1.13. Compare the $R^2$ score of the the linear fit with *medv* as the response and *lstat* as the predictor, diving the dataset into test and train, with a ratio of $30\%$ test dataset.

```python
# Establish the idxs of each test and train datasets
test_size = 0.3  # Ratio of test dataset
seed = 100  # Random seed
random.seed(seed)  # Apply random seed
idxs_test = random.sample(range(df.shape[0]), int(test_size * df.shape[0]))  # Sample idxs for test datasets
idxs_train = [i for i in range(df.shape[0]) if i not in idxs_test]  # Obtain the remaining idxs as training

# Assign the base input and output vectors
X = df['lstat'].values.reshape(-1, 1)
y = df['medv'].values.reshape(-1, 1)

# Apply the polynomial transformation, for polynomial degrees of 2 and 10
poly_2 = sklearn.preprocessing.PolynomialFeatures(degree=2, interaction_only=False, include_bias=False)
X_2 = poly_2.fit_transform(X)
poly_10 = sklearn.preprocessing.PolynomialFeatures(degree=10, interaction_only=False, include_bias=False)
X_10 = poly_10.fit_transform(X)

# Divide the input vectors in test and train datasets, using the idxs obtained
X_train_2 = X_2[idxs_train, :]
X_train_10 = X_10[idxs_train, :]
X_test_2 = X_2[idxs_test, :]
X_test_10 = X_10[idxs_test, :]

# Do the same for the output vector
y_train = y[idxs_train, :]
y_test = y[idxs_test, :]

# Fit the 2nd degree model to the data and predict the values of the test dataset
regr_2 = sklearn.linear_model.LinearRegression()
regr_2.fit(X_train_2, y_train)
medv_pred_2 = regr_2.predict(X_test_2)
r2_value_2 = sklearn.metrics.r2_score(y_test, medv_pred_2)

# Fit the 10th degree model to the data and predict the values of the test dataset
regr_10 = sklearn.linear_model.LinearRegression()
regr_10.fit(X_train_10, y_train)
medv_pred_10 = regr_10.predict(X_test_10)
r2_value_10 = sklearn.metrics.r2_score(y_test, medv_pred_10)

# Print results
print('R^2 (2nd degree): {:.3f}'.format(r2_value_2))
print('R^2 (10th degree): {:.3f}'.format(r2_value_10))

# Obtain sequence of 'lstat' points between the min and max values, and obtain the corresponding predicted values
# for both regression (2nd and 10th degrees)
lstat_vals = np.linspace(start=min(df['lstat']), stop=max(df['lstat']), num=1000)
medv_vals_10 = regr_10.predict(poly_10.fit_transform(lstat_vals.reshape(-1, 1)))
medv_vals_2 = regr_2.predict(poly_2.fit_transform(lstat_vals.reshape(-1, 1)))

# Plot the figure, showing the test vs train dataset and the 2nd vs 10th degree regressions

fig, ax = plt.subplots(figsize=(12,6))
ax.plot(X_train_2[:, 0], y_train[:, 0], '.r')
ax.plot(X_test_2[:, 0], y_test[:, 0], '.k')
ax.plot(lstat_vals, medv_vals_10, 'b-', label='10th degree')
ax.plot(lstat_vals, medv_vals_2, 'g-', label='2nd degree')
ax.set_xlim((0, 40))
ax.set_ylim((0, 55))
ax.legend()
```

#### 2.1. Load the file "carseats.csv" into a pandas dataframe.

```python
carseats_base = pd.read_csv("carseats.csv")
carseats_base.head()
```

#### 2.2. Pivot the column "ShelveLoc" into ShelveLocGood and ShelveLocMedium (bad comes by exclusion). Convert the "Urban" and "US" columns into numeric, making Yes = 1 and No = 0.

```python
carseats_base['ShelveLoc'].value_counts()
```

```python
# One-hot encoding
# Create columns
carseats = carseats_base.copy()

# Convert text data
carseats['ShelveLocGood'] = (carseats['ShelveLoc'] == 'Good')
carseats['ShelveLocMedium'] = (carseats['ShelveLoc'] == 'Medium')
carseats['Urban'] = (carseats['Urban'] == 'Yes')
carseats['US'] = (carseats['US'] == 'Yes')
carseats.drop(columns=['ShelveLoc'], inplace=True)
carseats.head()
```

Alternatively, we can use the function `pd.get_dummies()` for one-hot encoding.

```python
## Copy the base DataFrame
carseats2 = carseats_base.copy()

## Perform one-hot encoding using pd.get_dummies
carseats2 = pd.get_dummies(
    carseats2,
    columns=['ShelveLoc', 'Urban', 'US'],  # Specify the columns to one-hot encode
    drop_first=True  # Drop the first category
)
carseats2.head()
```

#### 2.3. Fit a linear model to the carseats data, considering the response as the Sales, and all other columns as predictors. Additionally consider the interaction elements "Income:Advertising" and "Price:Age"

```python
# Prepare the input data. Manually create the interaction elements' columns
X_df = carseats.drop(columns = 'Sales')
X_df['income:advertising'] = X_df['Income'] * X_df['Advertising']
X_df['price:age'] = X_df['Price'] * X_df['Age']
X = X_df.values
y = carseats['Sales'].values.reshape(-1, 1)

# Prepare the column names to be printed
names = list(X_df.columns.values)
names.insert(0, 'Intercept')

# Fit the linear regression model the data
regr = sklearn.linear_model.LinearRegression()
regr.fit(X, y)

# Extract the coefficients and print them
values = list(regr.coef_[0])
values.insert(0, regr.intercept_[0])
s = ['{} :: {:.4f}'.format(names[i].rjust(18), values[i]) for i in range(len(values))]
listToStr = '\n'.join([str(elem) for elem in s])
print(listToStr)

# Calculate and print the R^2 score
sales_pred = regr.predict(X)
r2_value = sklearn.metrics.r2_score(y, sales_pred)
print('\n{} :: {:.4f}'.format('R^2'.rjust(18), r2_value))
```
