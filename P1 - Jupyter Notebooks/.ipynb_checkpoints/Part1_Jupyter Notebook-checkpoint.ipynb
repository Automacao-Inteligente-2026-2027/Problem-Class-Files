## Environments

An environment is an isolated workspace that contains specific versions of Python, libraries, and tools needed for a project. It ensures that all dependencies are managed separately from the system-wide installation, allowing projects to run without interfering with each other and preventing errors when packages are updated.

In this course, it is recommended to create an environment to install Python and all the necessary packages. 
Whenever you run your code, you should run it within the environment you created.

### Miniconda 
[Miniconda](https://docs.conda.io/en/latest/miniconda.html) is a free minimal installer for conda, a package and environment manager. It is a small, bootstrap version of Anaconda that includes only conda, Python, the packages they depend on, and a small number of other useful packages, including pip, zlib and a few others. 


## Configuring your environment

### Creating your environment

With conda, you can create, export, list, remove, and update environments that have different versions of Python and/or packages installed in them. Switching or moving between environments is called activating the environment. You can also share an environment file.

To create an environment you should use the terminal or Anaconda Prompt and follow the steps below:

Create an environment:

> `conda create --name intelligent-automation`

Verify that the new environment was installed correctly:

> `conda env list`

Activate the new enviroment:

> `conda activate intelligent-automation`

Deactivate environment:
> `conda deactivate intelligent-automation`

For more details about enviroments access: [Managing environments](https://docs.conda.io/projects/conda/en/latest/user-guide/tasks/manage-environments.html)

### Installing Python
Activate again the environment you created with:
> `conda activate intelligent-automation`

Use `conda install` to install Python version 3.10 in the environment you just created. Installing this version of Python is important as more recent versions of Python are not compatible with some of the packages we will be using during the course. 

> `conda install python=3.10`

## Jupyter

[Jupyter](https://jupyter.org) is not strictly speaking an ML package. It provides a browser front-end connected to an instance of IPython which allows REPL (read-eval-print loop) for quick testing, allows to create documents that intertwines code, output, images, and text. This is great for prototyping, demonstrations and tutorials, but terrible for actual coding. 

[Jupyter Notebook](https://jupyter.org/install.html) is an open-source web application that allows you to create and share documents that contain live code, equations, visualizations and narrative text. Uses include: data cleaning and transformation, numerical simulation, statistical modeling, data visualization, machine learning, and much more.

[Jupyter Lab](https://jupyter.org/install.html) is a web-based interactive development environment for Jupyter notebooks, code, and data. JupyterLab is flexible: configure and arrange the user interface to support a wide range of workflows in data science, scientific computing, and machine learning. JupyterLab is extensible and modular: write plugins that add new components and integrate with existing ones.

Another way to install packages is through the [Python Package Index](https://pypi.org/) using the package `pip`. 
`pip` should be automatically installed when you install Python in the environment.

Make sure the environment is activated and install `jupyter` with the `pip install` command:
> `pip install jupyter`

Several more packages should be installed along with `jupyter`.
Run the command `pip  list` and make sure that `jupyter`, `notebook`, and `jupyterlab` are in the list of installed packages.



To start the Jupyter Notebook, make sure your environment is activated and run the following command: 
> `jupyter notebook`

A browser window should open with the Jupyter Notebook interface.

Similarly, for Jupyter Lab you can run the following command:
> `jupyter-lab`

### Read dataset file

```python
# Read from a .csv file
data_csv = pd.read_csv('iris.csv')

# Print the first 5 data points as the head of the dataset
data_csv.head(5)
```

```python
data_csv['sepal.length']*data_csv['sepal.width']
```

```python
# Read from a url 
url = "http://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"

names = ['sepal.length', 'sepal.width', 'petal.length', 'petal.width', 'variety']
data_url = pd.read_csv(url, names=names)

data_url.head(5)
```

### Examine data

```python
iris = data_csv

# Check values below 4.9 of the first 20 elements
test_iris = data_csv[0:20]

speal_length_double = test_iris['sepal.length']*2
speal_length_double
```

```python
#Statistical properties of the iris dataset
iris.describe()
```

```python
# Determine class distribution
print(iris.groupby('variety').size())
```

### Data Visualization

#### Matplotlib

```python
plt.rcParams['figure.figsize'] = [15, 6]
# Scatter plots
iris.plot(kind="scatter", x="petal.length", y="sepal.width")
```

```python
plt.rcParams['figure.figsize'] = [15, 6]
# Scatter plots
cmap = {'Setosa': "green", 'Versicolor': "red", 'Virginica': "blue"}
iris['color_column'] = iris['variety'].map(cmap)

iris.plot(kind="scatter", x="petal.length", y="sepal.width", c="color_column")
```

```python
iris.var
```

```python
# Scatter plot matrix
scatter_matrix(iris)
plt.show()
```

#### Seaborn

```python
# Color scatter plot by variety of iris dataset
sns.FacetGrid(iris, hue="variety", height=6) \
   .map(sns.scatterplot, "sepal.length", "sepal.width") \
   .add_legend()
```

```python
# Box plots adding a layer of individual points using stripplot
ax = sns.boxplot(x="variety", y="petal.length", hue='variety', data=iris)
ax = sns.stripplot(x="variety", y="petal.length", hue='variety', data=iris, jitter=True, edgecolor="auto")
```

```python
#  Pairplot shows the bivariate relation between each pair of features
# 
#  From the pairplot, it's possible to observe that the Iris-setosa species is separated from the other
#  two across all feature combinations

sns.pairplot(iris, hue="variety", height=3)
```

### Prepare data for training and validation

```python
# Convert pandas dataframe to numpy array since most scikit-learn libraries expect numpy arrays
array = iris.values

# Isolate Setosa variety
raw_target = array[:,4]
y = np.where(raw_target == 'Setosa', 1, 0)

print(y)
```

```python
#  Use all rows, columns 0, 1, 2, 3.
X = array[:,0:4]

#  Expect X to be 150 rows, 4 columns
print (type(X))
print ('Dimensions of X: ', X.shape)

#  Convert labels to 0/1 integers in the cell above
#  Expect y to be 150 rows, 1 column
print (type(y))
print ('Dimensions of y: ', y.shape)
```

```python
from sklearn.model_selection import train_test_split

test_size = 0.30
seed = 10
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=test_size, random_state=seed)
    
print ('Dimensions of X_train: ', X_train.shape)
print ('Dimensions of X_test:  ', X_test.shape)
print ('Dimensions of y_train: ', y_train.shape)
print ('Dimensions of y_test:  ', y_test.shape)
print ('Head of X_train: \n', X_train[:10])
print ('Head of X_test:  \n', X_test[:10])
print ('Head of y_train: ', y_train[:10])
print ('Head of y_test:  ', y_test[:10])
```

### Logistic Regression

```python
from sklearn.metrics import confusion_matrix
from sklearn.linear_model import LogisticRegression

# Apply a Logistic Regression model
model = LogisticRegression()
model.fit(X_train,y_train)
y_pred = model.predict(X_test)
cm = confusion_matrix(y_pred,y_test)

# Check confusion matrix
print ('                        Predicted')
print ('                   Positive','Negative')
print ('Actually Positive (1): ', cm[1,1], '    ', cm[1,0])
print ('Actually Negative (0): ', cm[0,1], '    ', cm[0,0])

print ('\nAll predictions are correct')
```
